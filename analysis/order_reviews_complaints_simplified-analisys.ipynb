A partir de agora estaremos analisando apenas a order_reviews_complaints_simplified.csv <br>Ela contém os dados necessários juntos para essa análise<br>Separei assim para melhor organização

```python
import matplotlib.pyplot as plt
import pandas as pd


df = pd.read_csv('created-tables/order_reviews_complaints_simplified.csv')
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])
df['order_delivered_customer_date'] = pd.to_datetime(df['order_delivered_customer_date'])
df['order_estimated_delivery_date'] = pd.to_datetime(df['order_estimated_delivery_date'])
```

```python
df["delivery_delay_days"] = (
    (df["order_delivered_customer_date"] - df["order_estimated_delivery_date"])
).dt.total_seconds() / 86400

df.to_csv('created-tables/order_reviews_complaints_simplified.csv')

df_entregas_atrasadas = df.loc[df['delivery_delay_days'] > 0]
media_de_atraso_por_nota = (
    df_entregas_atrasadas.groupby("review_score")
    .agg(
        qtd=('delivery_delay_days','mean')
    )
)
media_de_atraso_por_nota
```
