A partir de agora estaremos analisando apenas a order_reviews_complaints_simplified.csv (gerada no notebook score-initial-analisys)<br>
Ela contém todas as reviews com nota menor que 3 de pedidos entregues, com 1 linha por review

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv(
    "created-tables/order_reviews_complaints_simplified.csv",
    parse_dates=["order_purchase_timestamp", "order_delivered_customer_date", "order_estimated_delivery_date"],
)
df.shape
```

Entre as reclamações de pedidos **atrasados**, quantos dias de atraso em média para cada nota

```python
# delivery_delay_days já vem calculado do notebook score-initial-analisys
df_entregas_atrasadas = df.loc[df["entrega_atrasada"]]
atraso_por_nota = (
    df_entregas_atrasadas.groupby("review_score")
    .agg(
        qtd_pedidos=("review_id", "count"),
        media_dias_atraso=("delivery_delay_days", "mean"),
        mediana_dias_atraso=("delivery_delay_days", "median"),
    )
)
atraso_por_nota
```
